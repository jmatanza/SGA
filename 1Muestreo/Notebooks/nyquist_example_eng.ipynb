# Nyquist Sampling Explorer

Este simulador interactivo permite explorar el **teorema de muestreo de Nyquist–Shannon** y entender visualmente cuándo aparece **aliasing** al muestrear una señal continua.

Se analiza una señal sinusoidal:

$x(t) = \cos(2\pi f_1 t)$

que se muestrea con una frecuencia de muestreo $f_s$.

El objetivo del simulador es mostrar cómo el proceso de muestreo afecta al **espectro de la señal** y cuándo es posible **reconstruir la señal original**.

---

# Qué muestra el simulador

La figura contiene cuatro representaciones complementarias:

### 1️ Muestreo en el dominio temporal

Se muestran simultáneamente:

- la señal continua $x(t)$
- las muestras tomadas cada $T_s = 1/f_s$

Esto permite visualizar el proceso de discretización de la señal.

---

### 2️ Espectro de la señal

Se representa el espectro de:

- la señal original
- la señal muestreada

El muestreo genera **copias del espectro original**.

---

### 3️ Réplicas espectrales

Al muestrear una señal, su espectro se replica en múltiplos de la frecuencia de muestreo:

$X_s(f) = \sum_{k=-\infty}^{\infty} X(f - k f_s)$

La zona sombreada corresponde a la **primera zona de Nyquist** $[-f_s/2, f_s/2]$.

Si las réplicas espectrales **se solapan dentro de esta zona**, aparece **aliasing**.

---

### 4️ Reconstrucción de la señal

La señal se reconstruye mediante un **filtro paso bajo ideal** con frecuencia de corte:

$f_c = \frac{f_s}{2}$

Se puede observar:

- reconstrucción perfecta cuando no hay aliasing
- distorsión cuando el criterio de Nyquist no se cumple

---

# Parámetro interactivo

### Frecuencia de la señal $f_1$

El slider permite modificar la frecuencia de la señal original.

El comportamiento depende de la relación entre $f_1$ y la frecuencia de muestreo $f_s$:

- **$f_1 < f_s/2$** → muestreo correcto (sin aliasing)  
- **$f_1 > f_s/2$** → aparece aliasing  

---

## Respuestas

1. **Frecuencia máxima sin aliasing**

El teorema de Nyquist establece que no aparece aliasing si:

$f_{in} < \frac{F_s}{2}$

Por tanto, la frecuencia máxima de entrada sin aliasing es la **frecuencia de Nyquist**:

$f_{max} = \frac{F_s}{2}$

---

2. **Aliasing para una señal de 400 Hz**

Si la frecuencia de muestreo es $F_s = 300$ Hz, la frecuencia de Nyquist es:

$f_N = 150 \text{ Hz}$

La frecuencia alias observada es:

$f_{alias} = |f_{in} - kF_s|$

Para $f_{in} = 400$ Hz:

$f_{alias} = |400 - 300| = 100 \text{ Hz}$

Por tanto, la señal reconstruida tendrá **100 Hz**.

---

3. **Función del filtro paso bajo**

El filtro paso bajo del bloque reconstructor:

- elimina las **réplicas espectrales** generadas por el muestreo
- conserva únicamente la **primera zona de Nyquist**

Si no hay solapamiento espectral (no hay aliasing), este filtro permite **reconstruir exactamente la señal original**.

---

4. **Por qué el aliasing es irreversible**

Cuando ocurre aliasing, las **réplicas espectrales se solapan** en el dominio de la frecuencia.

Esto significa que:

- diferentes señales originales producen **exactamente las mismas muestras**
- la información original se pierde

Por tanto, **no existe forma de recuperar la señal original únicamente a partir de las muestras**.

---

5. **¿Se puede detectar aliasing observando la señal reconstruida?**

No necesariamente.

La señal reconstruida puede parecer una señal sinusoidal válida, pero con una frecuencia diferente.  
Sin conocer la señal original o la frecuencia de muestreo utilizada, **no es posible saber si ha habido aliasing**.

---

6. **Frecuencias que producen un alias de 10 Hz con Fs = 100 Hz**

El alias se obtiene mediante:

$f_{alias} = |f_{in} - kF_s|$

Si $F_s = 100$ Hz y queremos $f_{alias} = 10$ Hz:

Algunas posibles frecuencias de entrada son:

- $f_{in} = 10$ Hz  
- $f_{in} = 90$ Hz  
- $f_{in} = 110$ Hz  
- $f_{in} = 190$ Hz  
- $f_{in} = 210$ Hz  

En general:

$f_{in} = 10 + k \cdot 100$
o
$f_{in} = 90 + k \cdot 100$

donde $k$ es un número entero.
