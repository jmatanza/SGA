# Cuantificación de señales

En un sistema digital, las señales analógicas deben representarse mediante un número finito de niveles.  
Este proceso se denomina **cuantificación**.

La cuantificación introduce inevitablemente un **error**, denominado **error de cuantificación**.

## 1. Señal de prueba

La señal utilizada en esta simulación es la suma de dos componentes sinusoidales:

$x(t) = 0.9\cos(2\pi f_1 t) + 0.25\cos(2\pi f_2 t)$

donde:

- $f_1 = 120$ Hz
- $f_2 = 350$ Hz

La señal se normaliza para que su amplitud esté en el rango:

$-1 \le x(t) \le 1$

Esto permite utilizar un cuantificador uniforme con rango fijo.


## 2. Cuantificador uniforme

Un cuantificador uniforme divide el rango dinámico en **niveles igualmente espaciados**.

Si el número de bits es **B**, el número de niveles es:

$L = 2^B$

El tamaño del paso de cuantificación es:

$\Delta = \frac{x_{max}-x_{min}}{L-1}$

Cada muestra se aproxima al **nivel más cercano**.


## 3. Cuantificador uniforme

Un cuantificador uniforme divide el rango dinámico en **niveles igualmente espaciados**.

Si el número de bits es **B**, el número de niveles es:

$L = 2^B$

El tamaño del paso de cuantificación es:

$\Delta = \frac{x_{max}-x_{min}}{L-1}$

Cada muestra se aproxima al **nivel más cercano**.



## 4. Error de cuantificación

El error introducido por el cuantificador es:

$e(t) = x_q(t) - x(t)$

donde:

- $x(t)$ es la señal original
- $x_q(t)$ es la señal cuantificada

Para un cuantificador uniforme ideal:

$-\frac{\Delta}{2} \le e(t) \le \frac{\Delta}{2}$

El rendimiento del cuantificador se puede medir mediante el:

**Error cuadrático medio (MSE)**

$MSE = E[e^2(t)]$


## 5. Relación señal-ruido de cuantificación

Además de observar la señal cuantificada y el error, es útil analizar cómo cambia la **relación señal-ruido de cuantificación (SNR)** al variar el número de bits.

La SNR experimental puede calcularse como:

$\mathrm{SNR} = 10\log_{10}\left(\frac{P_x}{P_e}\right)$

donde:

- \(P_x\) es la potencia media de la señal
- \(P_e\) es la potencia media del error de cuantificación

Para una señal sinusoidal a escala completa, la teoría predice aproximadamente:

$\mathrm{SNR}_q \approx 6.02\,B + 1.76 \text{ dB}$

Esto significa que **cada bit adicional mejora la SNR en aproximadamente 6 dB**.
